# Titanic Exploratory Data Analysis (EDA)

## Project Questions

1. Did women survive more than men?
2. Did passenger ticket class affect survival?
3. Was age related to survival?

## Project Findings

1. Gender Analysis: Female passengers had a higher survival rate than male passengers.
2. Passenger Class Analysis: First-class passengers had a higher survival rate compared with second- and third-class passengers.
3. Age Analysis: Survival rates varied across different age groups.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df = sns.load_dataset("titanic")
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage)
```

```python
df["age"] = df["age"].fillna(
    df.groupby(["pclass", "sex"])["age"].transform("median")
)
```

```python
df = df.drop(columns=["deck"])
```

```python
df["embarked"] = df["embarked"].fillna(
    df["embarked"].mode()[0]
)
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nDataset shape:")
print(df.shape)
```

```python
df.groupby("sex")["survived"].mean().round(2)
```

```python
df.groupby("pclass")["survived"].agg(
    ["mean", "count"]
).round(2)
```

```python
df["age_band"] = pd.cut(
    df["age"],
    bins=[0, 12, 18, 40, 60, 100],
    labels=["Child", "Teen", "Adult", "Middle-aged", "Senior"]
)
```

```python
df.groupby("age_band", observed=True)["survived"].mean().round(2)
```

```python
print("SURVIVAL BY GENDER")
print(df.groupby("sex")["survived"].mean().round(2))

print("\nSURVIVAL BY PASSENGER CLASS")
print(df.groupby("pclass")["survived"].agg(
    ["mean", "count"]
).round(2))

print("\nSURVIVAL BY AGE GROUP")
print(df.groupby("age_band", observed=True)["survived"].mean().round(2))
```
